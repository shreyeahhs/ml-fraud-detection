# 05 – Inference Testing

In this notebook we:

- Load the trained model and feature artifacts.
- Run predictions on a small sample.
- Verify that the inference pipeline works end to end.


```
import pandas as pd

from src.predict import predict_proba, predict_labels
```

```
sample = [
    {"TransactionAmt": 100.0, "ProductCD": "W", "card1": 1000},
    {"TransactionAmt": 250.0, "ProductCD": "C", "card1": 2000},
]

sample_df = pd.DataFrame(sample)
sample_df
```

```
probs = predict_proba(sample_df)
labels = predict_labels(sample_df, threshold=0.5)

for i, row in sample_df.iterrows():
    print(f"Transaction {i}:")
    print(row.to_dict())
    print(f"Fraud probability: {probs[i]:.4f}, Predicted label: {labels[i]}")
    print()
```

## Result

We can now see:

- The full pipeline works on new data.
- The model outputs a probability between 0 and 1.
- We convert that to a label using a threshold.

This same logic is used in the REST API and in the batch inference script.
